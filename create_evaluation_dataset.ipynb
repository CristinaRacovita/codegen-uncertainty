# Evaluation Dataset Creation
To have a ground-truth dataset, I need to get each line from each problem in a different csv file to be easy to store all the data for them when doing the manual evaluation.

```python
import os
import pandas as pd
```

## Python Ground Truth Dataset

```python
python_csv_file = '/datasets/selected_problems_python_14B_no_comments_final.csv'
def get_generated_code(csv_file):
    csv_full_path = os.getcwd()+csv_file
    if os.path.exists(csv_full_path):
        return pd.read_csv(csv_full_path)
    else:
        print("File not found.")
        return None
    
python_solutions = get_generated_code(python_csv_file)
python_solutions.head()
```

```python
def create_empty_dataset():
    """Create an empty dataset with the required columns for code quality evaluation."""
    columns = [
        "problem_title",       # Title of the problem or source file
        "code_line",           # The actual line of code to evaluate
        "readability",         # 0/1 score for readability
        "naming",              # 0/1 score for naming conventions
        "length",              # 0/1 score for appropriate line length (auto-calculated)
        "logic",               # 0/1 score for logical complexity
        "formatting",          # 0/1 score for proper formatting
        "side_effects",        # 0/1 score for avoiding unexpected side effects
        "dry",                 # 0/1 score for not repeating code (Don't Repeat Yourself)
        "language_practice",   # 0/1 score for following language-specific practices
        "good"                 # 0/1 score based on the above criteria sum
    ]
    
    return pd.DataFrame(columns=columns)
```

```python
python_evaluation = create_empty_dataset()
```

```python
def get_code_line(code_solution):
    lines = code_solution.strip().split('\n')

    filtered_lines = []
    for line in lines:
        stripped = line.strip()
        filtered_lines.append(line)

    return filtered_lines
```

```python
def populate_code_line(solutions_df, python_evaluation):
    for _, row in solutions_df.iterrows():
        problem = row["title"]
        code = row["solution_no_comments"]
        
        lines = get_code_line(code)
        
        for line_num, line in enumerate(lines, 1):
            python_evaluation.loc[len(python_evaluation)] = [
                problem,
                line,
                None,
                None,
                1 if len(line) <= 80 else 0,
                None,
                None,
                None,
                None,
                None,
                None
            ]
populate_code_line(python_solutions, python_evaluation)
```
