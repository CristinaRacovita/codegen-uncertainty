## Ruff - Python Linter
I need to run the **Ruff Linter** for each solutions from *solutions/python* directory. I will save the results in *solutions/python/results*.

```python
import os
import json
import sys
import subprocess
from pathlib import Path
from typing import List, Dict, Any, Optional
```

```python
files = list(Path('solutions/python').glob('*.py'))
file_paths = [str(file) for file in files]
ruff_args = [
    # Target Python version
    "--target-version=py311",

    # Select all rule sets from your TOML
    "--select=E,W,F,I,C90,N,D,UP,YTT,ANN,ASYNC,S,BLE,FBT,B,A,COM,C4,DTZ,T10,EM,EXE,FA,ICN,INP,PIE,T20,PT,Q,RET,SIM,SLF,TID,INT,ARG,PTH,ERA,PL,TRY,FLY,NPY,PERF,RUF",

    # Ignore specific rules
    "--ignore=ANN201,D103,T201,D203,D212,INP001,D100",

    # Output format
    "--output-format=json",

    # Line length
    "--line-length=88",

    # Preview mode for unstable/experimental rules
    "--preview",
]
```

```python
def run_ruff_on_files(file_paths, ruff_args, output_dir = "lint_results/python", format_type = "json"):
    os.makedirs(output_dir, exist_ok=True)
    results = {}
    
    for file_path in file_paths:
        file_name = os.path.basename(file_path)
        output_file = os.path.join(output_dir, f"{file_name}.{format_type}")
        
        cmd = [sys.executable, "-m", "ruff", "check", file_path] + ruff_args
        
        try:
            result = subprocess.run(
                cmd, 
                capture_output=True, 
                text=True,
                check=False
            )
            
            with open(output_file, 'w') as f:
                f.write(result.stdout)
                
            if result.stderr:
                error_file = os.path.join(output_dir, f"{file_name}.error.txt")
                with open(error_file, 'w') as f:
                    f.write(result.stderr)
            
            results[file_path] = output_file
            print(f"Linted {file_path} -> {output_file}")
            
        except Exception as e:
            print(f"Error linting {file_path}: {e}")
    
    return results

run_ruff_on_files(file_paths, ruff_args)
```
