# Code Files Extraction
I want to have for each generated solution a different code file, with the correct extension. This way, I can run the linter over each of them and extract afterwards line quality features.

```python
import re
import os
import ast
import pandas as pd
from enum import Enum
```

```python
class Language(Enum):
    PYTHON = "python"
    TYPESCRIPT = "typescript"
    C_SHARP = "csharp"
```

```python
python_csv_file = '/datasets/selected_problems_python_14B_no_comments_final.csv'
c_sharp_csv_file = '/datasets/selected_problems_c_sharp_14B_no_comments_final.csv'
typescript_csv_file = '/datasets/selected_problems_typescript_14B_no_comments_final.csv'

def get_generated_code(csv_file):
    csv_full_path = os.getcwd()+csv_file
    if os.path.exists(csv_full_path):
        return pd.read_csv(csv_full_path)
    else:
        print("File not found.")
        return None
    
python_solutions = get_generated_code(python_csv_file)
c_sharp_solutions = get_generated_code(c_sharp_csv_file)
typescript_solutions = get_generated_code(typescript_csv_file)
```

```python
python_solutions.head()
```

```python
def extract_code_files(solutions, extension, language):
    for index, row in solutions.iterrows():
        with open(f"solutions/{language}/{row['title'].replace(' ','')}.{extension}", "w") as f:
            f.write(row['solution_no_comments'])
       
                
extract_code_files(python_solutions, 'py', 'python')
extract_code_files(c_sharp_solutions, 'cs', 'csharp')
extract_code_files(typescript_solutions, 'ts', 'typescript')
```
